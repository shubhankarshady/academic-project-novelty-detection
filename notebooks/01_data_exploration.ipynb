```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/raw")

for file in DATA_DIR.glob("*.xlsx"):
    print("\nFILE:", file.name)

    excel = pd.ExcelFile(file)

    print("Sheets:")
    for sheet in excel.sheet_names:
        df = pd.read_excel(file, sheet_name=sheet)
        print(f"  {sheet}: {df.shape}")
```

```python
DATA_DIR = Path("../data/raw")

file = DATA_DIR / "Academic_Project_Similarity_Synthetic_Dataset.xlsx"

projects = pd.read_excel(
    file,
    sheet_name="Projects"
)

pairs = pd.read_excel(
    file,
    sheet_name="SimilarityPairs"
)

dictionary = pd.read_excel(
    file,
    sheet_name="DataDictionary"
)

print("PROJECTS")
print(projects.columns.tolist())

print("\nSIMILARITY PAIRS")
print(pairs.columns.tolist())

print("\nDATA DICTIONARY")
display(dictionary)
```
